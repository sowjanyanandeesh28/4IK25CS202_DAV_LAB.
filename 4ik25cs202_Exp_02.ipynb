```python
import pandas as pd;
import numpy as np;
```

```python
data = {
    "Name": ["John", "Anna", "Peter", "Linda","Shweta","Shassy"],
    "Age": [25, np.nan, 30, 25, 28,22],
    "Marks": [85,90,np.nan,85,78,95],
    "City": [np.nan, "Paris", "Berlin", "London","Mysuru","Mysuru"]
}
df = pd.DataFrame(data);
```

```python
print("Dataset: ")
print(df)
```

```python
print("\n Shape of dataset(rows,columns): ",df.shape)
print("\nColumn name: ",df.columns.tolist())
print("\nData types of each column: ")
print(df.dtypes)
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive Statistics (numerical columns): ")
print(df.describe())
print("\nMean: \n",df.mean(numeric_only=True))
print("\nMedian: \n",df.median(numeric_only=True))
print("\nStandard Deviation: \n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset: ",df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in a dataset: ")
print(df[df.duplicated()])
print("\nNumber of duplicate rows in dataset: ",df.duplicated().sum())
print("\nEDA process completed successfully!!:)")
```
